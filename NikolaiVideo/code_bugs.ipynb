```python
import numpy as np
import cv2
import glob
import imutils
```

```python
image_paths = glob.glob('./images_bugs/*.jpeg')
images = []

for index, image in enumerate(image_paths):
    img = cv2.imread(image)
    images.append(img)
    # cv2.imshow(f"Image {index+1}", img)
    # cv2.waitKey(0)
```

```python
imageStitcher = cv2.Stitcher_create()

error, stitched_image = imageStitcher.stitch(images)

if not error:
    cv2.imwrite("BUGS_stitchedImage.png", stitched_image)
    # cv2.imshow("Final image", stitched_image)
```
